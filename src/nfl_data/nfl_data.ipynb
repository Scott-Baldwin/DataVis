Demo for pulling nfl game data

```python
import nfl_data_py as nfl
# import pandas as pd


def get_nfl_playoff_data(years):
    """
    Fetches schedule data and filters for playoff games.
    :param years: List of integers (e.g., [2023, 2024])
    """
    print(f"Fetching data for seasons: {years}...")

    # 1. Import schedules (includes regular season and playoffs)
    schedules = nfl.import_schedules(years)

    # 2. Filter for Postseason games
    # 'POST' stands for postseason; 'REG' for regular season
    playoff_games = schedules[schedules["game_type"] != "REG"].copy()

    # 3. Clean up the output (selecting most relevant columns)
    cols_to_keep = [
        "season",
        "week",
        "game_type",
        "gameday",
        "home_team",
        "away_team",
        "home_score",
        "away_score",
        "result",
        "total",
        "stadium",
    ]

    # Check if columns exist (sometimes schemas change slightly)
    available_cols = [c for c in cols_to_keep if c in playoff_games.columns]

    return playoff_games[available_cols]


# Example Usage: Get data for the 2023 and 2024 seasons
years_to_check = [2023, 2024]
df_playoffs = get_nfl_playoff_data(years_to_check)

# Display the first few playoff games
print(df_playoffs.head())

# Optional: Save to CSV
# df_playoffs.to_csv("nfl_playoffs_data.csv", index=False)
```
