```python
import polars as pl

df_pl = pl.read_csv("../data/nyc-tmy-2023.csv", skip_rows=2)
df_pl.group_by(["Year", "Month", "Day"], maintain_order=True).agg(
    pl.col("Temperature").mean()
)
```

```python
import pandas as pd

df_pd = pd.read_csv("../data/nyc-tmy-2023.csv", skiprows=2)
df_pd.groupby(["Year", "Month", "Day"]).mean().sort_values(by=["Month", "Day"])[
    "Temperature"
]
```
